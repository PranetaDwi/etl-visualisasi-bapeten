```python
def srp_zra(file_olah, provinsi="Semua", status_sumber="Semua"):

    kondisi = file_olah['jenis_sumber'] == "ZRA"
    if provinsi != "Semua":
        kondisi &= (file_olah["propinsi"] == provinsi)

    if status_sumber != "Semua":
        kondisi &= (file_olah["status_sumber"] == status_sumber)
        
    return len(file_olah.loc[kondisi])

srp_zra("Semua", "Non Aktif")
```

```python
import numpy as np

def srp_zra_prp(file_olah, tipe, provinsi="Semua", status_sumber="Semua"):
    # Filter utama jenis_sumber
    kondisi = file_olah['jenis_sumber'] == tipe

    # Handle provinsi (bisa berupa string tunggal atau list dari widget)
    if isinstance(provinsi, (list, tuple, np.ndarray)):
        if len(provinsi) > 0:
            kondisi &= file_olah["propinsi"].isin(provinsi)
    elif provinsi != "Semua":
        kondisi &= (file_olah["propinsi"] == provinsi)

    # Handle status_sumber (bisa berupa string tunggal atau list dari widget)
    if isinstance(status_sumber, (list, tuple, np.ndarray)):
        if len(status_sumber) > 0:
            kondisi &= file_olah["status_sumber"].isin(status_sumber)
    elif status_sumber != "Semua":
        kondisi &= (file_olah["status_sumber"] == status_sumber)
        
    return len(file_olah.loc[kondisi])


srp_zra_prp(file_df, "ZRA", "Banten", "Aktif")
```
